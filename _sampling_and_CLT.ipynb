```python
# 4A) Calculate SE
import math

def calculate_se(sigma, n):
    return sigma / math.sqrt(n)

sigma = 20

# Case 1: n = 100
n_100 = 100
se_100 = calculate_se(sigma, n_100)
print(f"For n = 100, SE = {se_100}")

# Case 2: n = 25
n_25 = 25
se_25 = calculate_se(sigma, n_25)
print(f"For n = 25, SE = {se_25}")

# b) Comparison
print(f"Comparison: As the sample size increases from 25 to 100, the Standard Error decreases from {se_25} to {se_100}.")
```

```python
#6A)Modify the simulation

#Replace sample_sizes with [2, 10, 50, 200].
#Keep exponential population
 #mean and SD equal to 10.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
sample_sizes = [2, 10, 50, 200]
repetitions = 10000

for n in sample_sizes:
    sample_means = []
    for i in range(repetitions):
        data = np.random.exponential(scale=10, size=n)
        sample_means.append(np.mean(data))

    simulated_se = np.std(sample_means)
    theoretical_se = 10 / np.sqrt(n)

    print("Sample Size:", n)
    print("Mean:", np.mean(sample_means))
    print("Simulated SE:", simulated_se)
    print("Theoretical SE:", theoretical_se)

    plt.hist(sample_means, bins=30)
    plt.title(f"Sampling Distribution (n={n})")
    plt.xlabel("Sample Means")
    plt.ylabel("Frequency")
    plt.show()
```
